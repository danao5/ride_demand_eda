```python
df_clean["pickup_hour"] = (
    df_clean["tpep_pickup_datetime"].dt.hour
)
```

```python
df_clean["pickup_day"] = (
    df_clean["tpep_pickup_datetime"].dt.day_name()
)
```

```python
df_clean["trip_duration_min"] = (
    (df_clean["tpep_dropoff_datetime"] - df_clean["tpep_pickup_datetime"]).dt.total_seconds()/60
)
```

```python
df_clean["avg_speed"] = (
    df_clean["trip_distance"] / (df_clean["trip_duration_min"] /60)
)
```

```python
hourly_demand = (
    df_clean.groupby("pickup_hour").size().reset_index(name="trip_count")
)

hourly_demand
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    hourly_demand["pickup_hour"],
    hourly_demand["trip_count"],
    marker="o"
)

plt.title("Ride Demand by Hour of the Day")
plt.xlabel("Hour")
plt.ylabel("Number of Trips")

plt.grid(True)
plt.show()
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

daily = (
    df_clean["pickup_day"].value_counts()
    .reindex(day_order)
)
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    daily.index,
    daily.values
)

plt.title("Ride Demand by Day of the Week", fontsize=14)
plt.xlabel("Day of Week")
plt.ylabel("Number of Rides")

plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
hourly_duration = (
    df_clean.groupby("pickup_hour")["trip_duration_min"].mean().reset_index()
)

hourly_duration.head()
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    hourly_duration["pickup_hour"],
    hourly_duration["trip_duration_min"],
    marker="o",
    linewidth=2
)

plt.title("Average Trip Duration by Pickup Hour", fontsize=14)
plt.xlabel("Hour of Day")
plt.ylabel("Average Trip Duration (minutes)")

plt.xticks(range(24))
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
df_clean["day_type"] = df_clean["pickup_day"].apply(
    lambda x: "Weekend" if x in ["Saturday", "Sunday"] else "Weekday"
)
```

```python
daytype = (
    df_clean["day_type"]
    .value_counts()
)
```

```python
plt.figure(figsize=(6,5))

plt.bar(
    daytype.index,
    daytype.values,
    color=['#1f77b4', '#ff7f0e']
)

plt.title("Weekday vs Weekend Ride Demand")
plt.xlabel("Day Type")
plt.ylabel("Number of Trips")

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
weekday_avg = daily.loc[
    ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
].mean()

weekend_avg = daily.loc[
    ["Saturday", "Sunday"]
].mean()

comparison = {
    "Weekday": weekday_avg,
    "Weekend": weekend_avg
}
```

```python
plt.figure(figsize=(6,5))

plt.bar(
    comparison.keys(),
    comparison.values(),
    color=['#1f77b4', '#ff7f0e']
)

plt.title("Average Daily Ride Demand")
plt.ylabel("Average Trips per Day")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
peak = (
    df_clean.groupby("pickup_hour")
    .size()
    .reset_index(name="Trips")
    .sort_values(
        by="Trips",
        ascending=False
    )
)

peak.head(5)
```

```python
plt.figure(figsize=(8,5))

top5 = peak.head(5)

plt.bar(
    top5["pickup_hour"].astype(str),
    top5["Trips"]
)

plt.title("Top 5 Peak Pickup Hours")
plt.xlabel("Hour")
plt.ylabel("Trips")

plt.grid(axis="y", alpha=0.3)

plt.show()
```
